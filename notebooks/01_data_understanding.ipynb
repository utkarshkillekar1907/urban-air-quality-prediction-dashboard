```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
cpcb_2024 = pd.read_csv("../data/raw/CPCB/Raw_Data_2024_site_5792_ambazari_nagpur_mpcb_1Hr.csv")

cpcb_2024.sample(5)
```

```python
cpcb_2025 = pd.read_csv("../data/raw/CPCB/Raw_Data_2025_site_5792_ambazari_nagpur_mpcb_1Hr.csv")

cpcb_2025.sample(5)
```

```python
weather = pd.read_csv("../data/raw/Weather/open-meteo-21.12N79.02E323m.csv", skiprows=2)

weather.sample(5)
```

```python
print("CPCB 2024:", cpcb_2024.shape)
print("CPCB 2025:", cpcb_2025.shape)
print("Weather:", weather.shape)
```

```python
print("CPCB 2024 columns:")
print(cpcb_2024.columns.tolist())

print("\nCPCB 2025 columns:")
print(cpcb_2025.columns.tolist())

print("\nWeather columns:")
print(weather.columns.tolist())
```

```python
print("CPCB 2024:")
print(cpcb_2024.dtypes)

print("\nWeather:")
print(weather.dtypes)
```

```python
print("CPCB 2024 missing values:")
print(cpcb_2024.isnull().sum().sort_values(ascending=False))
```

```python
print("\nCPCB 2025 missing values:")
print(cpcb_2025.isnull().sum().sort_values(ascending=False))
```

```python
print("\nWeather missing values:")
print(weather.isnull().sum())
```

```python
cpcb_2024[cpcb_2024["PM2.5 (µg/m³)"].isnull()][["Timestamp", "PM2.5 (µg/m³)"]].head(20)
```

```python
cpcb_2025[cpcb_2025["PM2.5 (µg/m³)"].isnull()][["Timestamp", "PM2.5 (µg/m³)"]].head(20)
```

```python
target_pollutants = ["PM2.5 (µg/m³)","PM10 (µg/m³)","NO2 (µg/m³)","SO2 (µg/m³)","CO (mg/m³)","Ozone (µg/m³)"]

missing_2024 = cpcb_2024[target_pollutants].isnull().mean() * 100
missing_2025 = cpcb_2025[target_pollutants].isnull().mean() * 100

print("2024 Missing Percentage:")
print(missing_2024.round(2))

print("\n2025 Missing Percentage:")
print(missing_2025.round(2))
```

```python
print("CPCB 2024 duplicate timestamps:",
      cpcb_2024["Timestamp"].duplicated().sum())

print("CPCB 2025 duplicate timestamps:",
      cpcb_2025["Timestamp"].duplicated().sum())

print("Weather duplicate timestamps:",
      weather["time"].duplicated().sum())
```

```python
cpcb_2024["Timestamp"] = pd.to_datetime(cpcb_2024["Timestamp"])
cpcb_2025["Timestamp"] = pd.to_datetime(cpcb_2025["Timestamp"])
weather["time"] = pd.to_datetime(weather["time"])

print(cpcb_2024["Timestamp"].dtype)
print(cpcb_2025["Timestamp"].dtype)
print(weather["time"].dtype)
```

```python
print("CPCB 2024:")
print(cpcb_2024["Timestamp"].min())
print(cpcb_2024["Timestamp"].max())

print("\nCPCB 2025:")
print(cpcb_2025["Timestamp"].min())
print(cpcb_2025["Timestamp"].max())

print("\nWeather:")
print(weather["time"].min())
print(weather["time"].max())
```
