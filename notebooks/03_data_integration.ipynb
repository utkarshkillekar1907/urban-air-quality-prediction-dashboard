```python
import pandas as pd
import numpy as np
```

```python
cpcb_2024 = pd.read_csv("../data/processed/CPCB_Ambazari_2024_cleaned.csv")

cpcb_2025 = pd.read_csv("data/processed/CPCB_Ambazari_2025_cleaned.csv")

weather = pd.read_csv("../data/processed/OpenMeteo_Nagpur_2024_2025_cleaned.csv")
```

```python
cpcb_2024["timestamp"] = pd.to_datetime(cpcb_2024["timestamp"])
cpcb_2025["timestamp"] = pd.to_datetime(cpcb_2025["timestamp"])
weather["timestamp"] = pd.to_datetime(weather["timestamp"])
```

```python
print("CPCB 2024:", cpcb_2024.shape)
print("CPCB 2025:", cpcb_2025.shape)
print("Weather:", weather.shape)
```

```python
print("CPCB 2024:")
print(cpcb_2024["timestamp"].min(), "→", cpcb_2024["timestamp"].max())

print("\nCPCB 2025:")
print(cpcb_2025["timestamp"].min(), "→", cpcb_2025["timestamp"].max())

print("\nWeather:")
print(weather["timestamp"].min(), "→", weather["timestamp"].max())
```

```python
cpcb_all = pd.concat([cpcb_2024, cpcb_2025], ignore_index=True)

cpcb_all = cpcb_all.sort_values("timestamp").reset_index(drop=True)

print(cpcb_all.shape)
print(cpcb_all["timestamp"].duplicated().sum())
```

```python
integrated_data = pd.merge(cpcb_all, weather, on="timestamp", how="inner")

print("Integrated dataset shape:", integrated_data.shape)
```

```python
integrated_data = integrated_data.rename(columns={
    "wind_speed_x": "cpcb_wind_speed",
    "temperature": "cpcb_temperature",
    "humidity": "cpcb_humidity",
    "wind_direction": "cpcb_wind_direction",
    "rainfall": "cpcb_rainfall",
    "solar_radiation": "cpcb_solar_radiation",
    "temperature_weather": "weather_temperature",
    "humidity_weather": "weather_humidity",
    "wind_direction_weather": "weather_wind_direction",
    "wind_speed_y": "weather_wind_speed"
})

print(integrated_data.head(2).T)
print("\nColumns:")
print(integrated_data.columns.tolist())
```

```python
integrated_data = integrated_data[
    [
        "timestamp",
        "pm25",
        "pm10",
        "no2",
        "so2",
        "co",
        "o3",
        "weather_temperature",
        "weather_humidity",
        "weather_wind_speed",
        "weather_wind_direction",
        "weather_code"
    ]
].copy()
```

```python
print("Shape:", integrated_data.shape)

print("\nMissing values:")
print(integrated_data.isnull().sum())

print("\nDuplicate timestamps:",
      integrated_data["timestamp"].duplicated().sum())

print("\nDate range:")
print(
    integrated_data["timestamp"].min(),
    "→",
    integrated_data["timestamp"].max()
)
```

```python
integrated_data = integrated_data.rename(columns={
    "weather_temperature": "temperature",
    "weather_humidity": "humidity",
    "weather_wind_speed": "wind_speed",
    "weather_wind_direction": "wind_direction"
})
```

```python
print("Final shape:", integrated_data.shape)

print("\nFinal columns:")
print(integrated_data.columns.tolist())

print("\nMissing values:")
print(integrated_data.isnull().sum())
```

```python
integrated_data.to_csv("../data/processed/Ambazari_Nagpur_Integrated_2024_2025.csv", index=False)

print("Integrated dataset saved successfully.")
```
