```python
cpcb_2024 = pd.read_csv("../data/raw/CPCB/Raw_Data_2024_site_5792_ambazari_nagpur_mpcb_1Hr.csv")

cpcb_2025 = pd.read_csv("../data/raw/CPCB/Raw_Data_2025_site_5792_ambazari_nagpur_mpcb_1Hr.csv")

weather = pd.read_csv("../data/raw/Weather/open-meteo-21.12N79.02E323m.csv", skiprows=2)
```

```python
cpcb_2024["Timestamp"] = pd.to_datetime(cpcb_2024["Timestamp"])
cpcb_2025["Timestamp"] = pd.to_datetime(cpcb_2025["Timestamp"])
weather["time"] = pd.to_datetime(weather["time"])
```

```python
cpcb_columns = [
    "Timestamp",
    "PM2.5 (µg/m³)",
    "PM10 (µg/m³)",
    "NO2 (µg/m³)",
    "SO2 (µg/m³)",
    "CO (mg/m³)",
    "Ozone (µg/m³)",
    "AT (°C)",
    "RH (%)",
    "WS (m/s)",
    "WD (deg)",
    "RF (mm)",
    "SR (W/mt2)"
]

cpcb_2024_clean = cpcb_2024[cpcb_columns].copy()
cpcb_2025_clean = cpcb_2025[cpcb_columns].copy()
```

```python
target_pollutants = ["pm25", "pm10", "no2", "so2", "co", "o3"]

imputation_data = cpcb_2024_clean[target_pollutants].copy()

# Keep only rows where all six pollutants are originally available
complete_data = imputation_data.dropna().copy()

print("Original CPCB 2024 rows:", len(imputation_data))
print("Complete rows available for experiment:", len(complete_data))
```

```python
from sklearn.model_selection import train_test_split

# Make a copy of the complete data
validation_data = complete_data.copy()

# Randomly hide 10% of the values
rng = np.random.default_rng(42)

mask = rng.random(validation_data.shape) < 0.10

masked_data = validation_data.mask(mask)

print("Total values:", validation_data.size)
print("Artificially hidden values:", mask.sum())
print("Remaining known values:", masked_data.notna().sum().sum())
```

```python
mean_imputed = masked_data.copy()

for column in target_pollutants:
    mean_imputed[column] = mean_imputed[column].fillna(masked_data[column].mean())
```

```python
median_imputed = masked_data.copy()

for column in target_pollutants:
    median_imputed[column] = median_imputed[column].fillna(masked_data[column].median())
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error

results = []

for column in target_pollutants:

    # Only evaluate the values that we intentionally masked
    masked_positions = mask[:, target_pollutants.index(column)]

    actual = validation_data.loc[masked_positions, column]
    mean_pred = mean_imputed.loc[masked_positions, column]
    median_pred = median_imputed.loc[masked_positions, column]

    results.append({
        "pollutant": column,
        "mean_MAE": mean_absolute_error(actual, mean_pred),
        "mean_RMSE": np.sqrt(mean_squared_error(actual, mean_pred)),
        "median_MAE": mean_absolute_error(actual, median_pred),
        "median_RMSE": np.sqrt(mean_squared_error(actual, median_pred))
    })

results_df = pd.DataFrame(results)

results_df
```

```python
from sklearn.impute import KNNImputer

# Standardize the data because the pollutants have different scales
column_means = masked_data.mean()
column_stds = masked_data.std()

scaled_data = (masked_data - column_means) / column_stds

# KNN imputation
knn_imputer = KNNImputer(n_neighbors=5, weights="distance")

knn_array = knn_imputer.fit_transform(scaled_data)

# Convert back to DataFrame
knn_imputed_scaled = pd.DataFrame(
    knn_array,
    columns=target_pollutants,
    index=masked_data.index
)

# Convert back to original scale
knn_imputed = (
    knn_imputed_scaled * column_stds
) + column_means

knn_imputed.head()
```

```python
knn_results = []

for column in target_pollutants:

    # Positions that were intentionally masked
    masked_positions = mask[:, target_pollutants.index(column)]

    actual = validation_data.loc[validation_data.index[masked_positions], column]
    knn_pred = knn_imputed.loc[validation_data.index[masked_positions], column]

    knn_results.append({
        "pollutant": column,
        "knn_MAE": mean_absolute_error(actual, knn_pred),
        "knn_RMSE": np.sqrt(mean_squared_error(actual, knn_pred))
    })

knn_results_df = pd.DataFrame(knn_results)

knn_results_df
```

```python
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer

# Create MICE imputer
mice_imputer = IterativeImputer(
    random_state=42,
    max_iter=10
)

# Perform MICE imputation
mice_array = mice_imputer.fit_transform(masked_data)

# Convert to DataFrame
mice_imputed = pd.DataFrame(
    mice_array,
    columns=target_pollutants,
    index=masked_data.index
)

mice_imputed.head()
```

```python
mice_results = []

for column in target_pollutants:

    # Positions that were intentionally masked
    masked_positions = mask[:, target_pollutants.index(column)]

    actual = validation_data.loc[
        validation_data.index[masked_positions], column
    ]

    mice_pred = mice_imputed.loc[
        validation_data.index[masked_positions], column
    ]

    mice_results.append({
        "pollutant": column,
        "mice_MAE": mean_absolute_error(actual, mice_pred),
        "mice_RMSE": np.sqrt(mean_squared_error(actual, mice_pred))
    })

mice_results_df = pd.DataFrame(mice_results)

mice_results_df
```

```python
#MICE (Multiple Imputation by Chained Equations) is selected as the primary imputation method for the pollutant variables because it achieved the lowest RMSE for all six pollutants and the lowest MAE for five of six pollutants in the controlled validation experiment.
final_comparison = results_df.merge(
    knn_results_df,
    on="pollutant"
).merge(
    mice_results_df,
    on="pollutant"
)

final_comparison
```
